### Задача 1: Декоратор `print_boundary`
Создайте декоратор `print_boundary`, который печатает строку `\"--- START ---\"` перед вызовом функции и `\"--- END ---\"` после её выполнения.

---

### Task 1: `print_boundary` Decorator
Create a decorator `print_boundary` that prints the string `\"--- START ---\"` before the function call and `\"--- END ---\"` after its execution.


```python
def print_boundary(func):
    def wrapper(*args, **kwargs):
        print("--- START ---")
        result = func(*args, **kwargs)
        print("--- END ---")
        return result
    return wrapper
```

### Задача 2: Декоратор `uppercase_result`
Напишите декоратор `uppercase_result`, который преобразует результат выполнения функции в верхний регистр. Декоратор должен работать только с функциями, возвращающими строки.

---

### Task 2: `uppercase_result` Decorator
Write a decorator `uppercase_result` that converts the result of a function to uppercase. The decorator should only work with functions that return strings.


```python
def uppercase_result(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        if isinstance(result, str):
            return result.upper()
        return result
    return wrapper
```

### Задача 3: Генератор `countdown`
Создайте функцию-генератор `countdown(n)`, которая "отсчитывает" числа от `n` до 0 включительно.

---

### Task 3: `countdown` Generator
Create a generator function `countdown(n)` that yields numbers from `n` down to 0, inclusive.


```python
def countdown(n):
    while n >= 0:
        yield n
        n -= 1
```

### Задача 4: Итератор `ReverseIterator`
Напишите класс-итератор `ReverseIterator`, который принимает строку и итерируется по её символам в обратном порядке.

---

### Task 4: `ReverseIterator` Iterator
Write an iterator class `ReverseIterator` that takes a string and iterates over its characters in reverse order.


```python
class ReverseIterator:
    def __init__(self, string):
        self.string = string
        self.index = len(string) - 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.index < 0:
            raise StopIteration
        char = self.string[self.index]
        self.index -= 1
        return char
```

### Task 5: `log_call` Decorator
Create a decorator `log_call` that logs (prints) the name of the called function, its positional and keyword arguments, and its return value.


```python
def log_call(func):
    def wrapper(*args, **kwargs):
        print(f"Function: {func.__name__}")
        print(f"Positional args: {args}")
        print(f"Keyword args: {kwargs}")
        result = func(*args, **kwargs)
        print(f"Return value: {result}")
        return result
    return wrapper
```

Напишите параметризованный декоратор `repeat(n)`, который заставляет декорируемую функцию выполняться `n` раз и возвращает список результатов всех вызовов.

---

### Task 6: Parameterized Decorator `repeat`
Write a parameterized decorator `repeat(n)` that causes the decorated function to be executed `n` times and returns a list of the results from all calls.


```python
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            return [func(*args, **kwargs) for _ in range(n)]
        return wrapper
    return decorator
```

### Задача 7: Генератор `file_lines_generator`
Напишите функцию-генератор `file_lines_generator(filepath)`, которая принимает путь к файлу и "отдает" (yields) каждую строку этого файла по одной. Генератор должен корректно обрабатывать ситуацию, когда файл не найден (например, с помощью `try...except FileNotFoundError`).

---

### Task 7: `file_lines_generator` Generator
Write a generator function `file_lines_generator(filepath)` that takes a file path and yields each line of that file one by one. The generator should correctly handle the case where the file is not found (e.g., using `try...except FileNotFoundError`).


```python
def file_lines_generator(filepath):
    try:
        with open(filepath, 'r', encoding='utf-8') as f:
            for line in f:
                yield line
    except FileNotFoundError:
        print(f"File not found: {filepath}")
```

### Задача 8: Генератор `prime_generator`
Создайте генератор `prime_generator(limit)`, который генерирует все простые числа до заданного `limit`.

---

### Task 8: `prime_generator` Generator
Create a generator `prime_generator(limit)` that generates all prime numbers up to a given `limit`.


```python
def prime_generator(limit):
    for num in range(2, limit + 1):
        is_prime = True
        for divisor in range(2, int(num ** 0.5) + 1):
            if num % divisor == 0:
                is_prime = False
                break
        if is_prime:
            yield num
```

При последующих вызовах с теми же аргументами должен возвращаться сохраненный результат без выполнения функции. (Подсказка: используйте словарь для хранения кэша).

### Task 9: `cache_result` Decorator
Implement a decorator `cache_result` that caches the result of a function's execution. On the first call with specific arguments, the function should run, and its result should be stored. On subsequent calls with the same arguments, the stored result should be returned without executing the function. (Hint: use a dictionary to store the cache).


```python
def cache_result(func):
    cache = {}

    def wrapper(*args, **kwargs):
        key = (args, tuple(sorted(kwargs.items())))
        if key not in cache:
            cache[key] = func(*args, **kwargs)
        return cache[key]

    return wrapper
```

### Задача 10: Генератор `nested_list_unpacker`
Напишите генератор `nested_list_unpacker(nested_list)`, который принимает вложенный список (любой глубины) и "распаковывает" его, возвращая все элементы в виде одной плоской последовательности. Например, для `[1, [2, 3], [4, [5, 6]]]` он должен вернуть `1, 2, 3, 4, 5, 6`.

---

### Task 10: `nested_list_unpacker` Generator
Write a generator `nested_list_unpacker(nested_list)` that takes a nested list (of any depth) and "unpacks" it, yielding all elements as a single flat sequence. For example, for `[1, [2, 3], [4, [5, 6]]]`, it should yield `1, 2, 3, 4, 5, 6`.


```python
def nested_list_unpacker(nested_list):
    for item in nested_list:
        if isinstance(item, list):
            yield from nested_list_unpacker(item)
        else:
            yield item
```

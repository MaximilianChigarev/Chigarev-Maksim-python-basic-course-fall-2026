### Task 2: `User` Class with Password Validation
Create a `User` class to represent a user.
*   **Attributes:** `username` (public), `__password` (private).
*   **Methods:**
    *   `__init__`: accepts `username` and `password`.
    *   `set_password(new_password)`: sets a new password. The password must be at least 8 characters long. If the password is shorter, raise a `ValueError`.
    *   `check_password(password_to_check)`: returns `True` if the provided password matches the user's password, and `False` otherwise.
    *   *Note:* There should be no direct way to "get" the password (`get_password`) to demonstrate encapsulation.


```python
class User:

    MIN_PASSWORD_LENGTH = 8

    def __init__(self, username: str, password: str):

        self.username = username
        self.set_password(password)

    def set_password(self, new_password: str) -> None:

        if len(new_password) < self.MIN_PASSWORD_LENGTH:
            raise ValueError(
                f"Пароль слишком короткий: минимум "
                f"{self.MIN_PASSWORD_LENGTH} символов, "
                f"получено {len(new_password)}."
            )
        self.__password = new_password

    def check_password(self, password_to_check: str) -> bool:

        return self.__password == password_to_check
```

Создайте класс-утилиту `StringUtils`, который будет содержать статические методы для работы со строками. Этот класс не должен иметь `__init__` или атрибутов экземпляра.

    *   `@staticmethod count_vowels(s)`: принимает строку `s` и возвращает количество гласных букв в ней (a, e, i, o, u, как строчных, так и прописных).

### Task 3: `StringUtils` Class with Static Methods
Create a utility class `StringUtils` that will contain static methods for working with strings. This class should not have an `__init__` method or instance attributes.
*   **Methods:**
    *   `@staticmethod reverse_string(s)`: takes a string `s` and returns it reversed.
    *   `@staticmethod is_palindrome(s)`: takes a string `s` and returns `True` if it is a palindrome (reads the same forwards and backwards), and `False` otherwise.
    *   `@staticmethod count_vowels(s)`: takes a string `s` and returns the number of vowels in it (a, e, i, o, u, both lowercase and uppercase).


```python
class StringUtils:

    @staticmethod
    def reverse_string(s: str) -> str:
        return s[::-1]

    @staticmethod
    def is_palindrome(s: str) -> bool:

        cleaned = s.lower().replace(" ", "")
        return cleaned == cleaned[::-1]

    @staticmethod
    def count_vowels(s: str) -> int:
        vowels = set("aeiouAEIOU")
        return sum(1 for char in s if char in vowels)
```

    *   Создайте свойство `is_adult` (только для чтения), которое возвращает `True`, если возраст 18 лет и больше, и `False` в противном случае.

### Task 4: `Person` Class with `age` Property
Create a `Person` class that stores a person's age.
*   **Attributes:** `_name` (protected), `_age` (protected).
*   **Properties:**
    *   Create an `age` property.
    *   The getter (`@property`) for `age` should return the value of `_age`.
    *   The setter (`@age.setter`) should check that the age being set is in the range of 0 to 120. If the age is invalid, raise a `ValueError`.
    *   Create a read-only property `is_adult` that returns `True` if the age is 18 or greater, and `False` otherwise.


```python
class Person:

    MIN_AGE = 0
    MAX_AGE = 120
    ADULT_AGE = 18

    def __init__(self, name: str, age: int):
        self._name = name
        self.age = age

    @property
    def age(self) -> int:
        return self._age

    @age.setter
    def age(self, value: int) -> None:
        if not isinstance(value, int) or isinstance(value, bool):
            raise ValueError(f"Возраст должен быть целым числом, получено: {value!r}")
        if not (self.MIN_AGE <= value <= self.MAX_AGE):
            raise ValueError(
                f"Возраст должен быть в диапазоне "
                f"{self.MIN_AGE}–{self.MAX_AGE}, получено: {value}"
            )
        self._age = value

    @property
    def is_adult(self) -> bool:
        return self._age >= self.ADULT_AGE
```

### Task 6: Notification System (Abstract Classes)
Create a notification system using abstract classes.
*   Create an abstract class `Notification` using the `abc` module. It should have one abstract method `send(message)`.
*   Create three concrete classes that inherit from `Notification`:
    1.  `EmailNotification`: accepts `recipient_email` in its constructor. The `send` method should print "Sending Email to [address]: [message]".
    2.  `SMSNotification`: accepts `phone_number` in its constructor. The `send` method should print "Sending SMS to [number]: [message]".
    3.  `PushNotification`: accepts `device_id` in its constructor. The `send` method should print "Sending Push notification to device [ID]: [message]".
*   Write a function that takes a list of notification objects and a message, and then calls the `send` method for each of them.


```python
from abc import ABC, abstractmethod


class Notification(ABC):
    @abstractmethod
    def send(self, message: str) -> None:
        ...


class EmailNotification(Notification):
    def __init__(self, recipient_email: str):
        self.recipient_email = recipient_email

    def send(self, message: str) -> None:
        print(f"Отправка Email на {self.recipient_email}: {message}")


class SMSNotification(Notification):
    def __init__(self, phone_number: str):
        self.phone_number = phone_number

    def send(self, message: str) -> None:
        print(f"Отправка SMS на {self.phone_number}: {message}")


class PushNotification(Notification):
    def __init__(self, device_id: str):
        self.device_id = device_id

    def send(self, message: str) -> None:
        print(f"Отправка Push-уведомления на устройство {self.device_id}: {message}")


def notify_all(notifications, message: str) -> None:
    for notification in notifications:
        notification.send(message)
```

### Task 7: Composition "Library and Books"
Create a library management system that demonstrates composition.
*   Use the `Book` class from **Task 1**.
*   Create a `Library` class that "has" a list of books.
*   **`Library` Attributes:** `name` (library name), `books` (a list to store `Book` objects).
*   **`Library` Methods:**
    *   `add_book(book)`: adds a book object to the list.
    *   `remove_book(title)`: removes a book by its title.
    *   `find_book_by_author(author_name)`: searches for and returns a list of all books by a given author.


```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def get_info(self):
        return f"Книга: {self.title}, Автор: {self.author}, Страниц: {self.pages}"

    def __str__(self):
        return self.get_info()


class Library:
    def __init__(self, name):
        self.name = name
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def remove_book(self, title):
        for book in self.books:
            if book.title == title:
                self.books.remove(book)
                return True
        return False

    def find_book_by_author(self, author_name):
        return [book for book in self.books if book.author == author_name]
```

### Task 8: Magic Methods for `Vector` Class
Create a `Vector` class to represent a 2D vector.
*   **Attributes:** `x`, `y`.
*   Implement the following magic methods:
    *   `__add__(self, other)`: for adding vectors (`v1 + v2`).
    *   `__sub__(self, other)`: for subtracting vectors (`v1 - v2`).
    *   `__mul__(self, scalar)`: for multiplying a vector by a scalar (number) (`v1 * 3`).
    *   `__str__` and `__repr__` for elegant output.


```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)

    def __str__(self):
        return f"Vector({self.x}, {self.y})"

    def __repr__(self):
        return f"Vector(x={self.x}, y={self.y})"
```

### Task 9: Task Manager System
Design and implement a console-based task manager.
*   **Classes:**
    *   `Task`: with attributes `description`, `due_date`, `status` (e.g., "pending", "in progress", "completed").
    *   `TaskManager`: manages a list of tasks.
*   **`TaskManager` Functionality:**
    *   Adding a new task.
    *   Viewing the list of all tasks.
    *   Changing a task's status by its description or ID.
    *   Deleting a task.
    *   Add the ability to save tasks to a file (e.g., JSON) and load them from it.


```python
import json


class Task:
    def __init__(self, description, due_date, status="в ожидании", task_id=None):
        self.id = task_id
        self.description = description
        self.due_date = due_date
        self.status = status

    def to_dict(self):
        return {
            "id": self.id,
            "description": self.description,
            "due_date": self.due_date,
            "status": self.status,
        }

    @classmethod
    def from_dict(cls, data):
        return cls(
            description=data["description"],
            due_date=data["due_date"],
            status=data["status"],
            task_id=data["id"],
        )

    def __str__(self):
        return f"[{self.id}] {self.description} | срок: {self.due_date} | статус: {self.status}"

    def __repr__(self):
        return f"Task(id={self.id!r}, description={self.description!r}, due_date={self.due_date!r}, status={self.status!r})"


class TaskManager:
    STATUSES = ("в ожидании", "в процессе", "завершено")

    def __init__(self, storage_file="tasks.json"):
        self.storage_file = storage_file
        self.tasks = []
        self._next_id = 1

    def add_task(self, description, due_date, status="в ожидании"):
        if status not in self.STATUSES:
            raise ValueError(f"Недопустимый статус: {status}. Допустимо: {self.STATUSES}")
        task = Task(description, due_date, status, self._next_id)
        self._next_id += 1
        self.tasks.append(task)
        return task

    def list_tasks(self):
        if not self.tasks:
            print("Список задач пуст.")
            return
        for task in self.tasks:
            print(task)

    def _find_task(self, key):
        for task in self.tasks:
            if str(task.id) == str(key) or task.description == key:
                return task
        return None

    def change_status(self, key, new_status):
        if new_status not in self.STATUSES:
            raise ValueError(f"Недопустимый статус: {new_status}. Допустимо: {self.STATUSES}")
        task = self._find_task(key)
        if task is None:
            print(f"Задача не найдена: {key}")
            return False
        task.status = new_status
        print(f"Статус задачи [{task.id}] изменён на «{new_status}».")
        return True

    def remove_task(self, key):
        task = self._find_task(key)
        if task is None:
            print(f"Задача не найдена: {key}")
            return False
        self.tasks.remove(task)
        print(f"Задача [{task.id}] удалена.")
        return True

    def save(self):
        data = [task.to_dict() for task in self.tasks]
        with open(self.storage_file, "w", encoding="utf-8") as f:
            json.dump(data, f, ensure_ascii=False, indent=2)
        print(f"Сохранено задач: {len(data)} → {self.storage_file}")

    def load(self):
        try:
            with open(self.storage_file, "r", encoding="utf-8") as f:
                data = json.load(f)
        except FileNotFoundError:
            print(f"Файл не найден: {self.storage_file}")
            return
        except json.JSONDecodeError:
            print(f"Файл повреждён: {self.storage_file}")
            return
        self.tasks = [Task.from_dict(item) for item in data]
        self._next_id = max((t.id for t in self.tasks), default=0) + 1
        print(f"Загружено задач: {len(self.tasks)} из {self.storage_file}")
```

### Task 10: "Factory" Pattern for Document Creation
Implement the "Factory Method" pattern to create different types of documents.
*   Create an abstract class `Document` with an abstract method `display()`.
*   Create concrete document classes: `PDFDocument`, `WordDocument`, `WebDocument`, each implementing the `display()` method (e.g., printing a string "Displaying PDF document").
*   Create a `DocumentFactory` class with a static method `create_document(doc_type)`. This method returns an instance of the corresponding document class based on a string (`"pdf"`, `"word"`, `"web"`).
*   The client code should only use the factory to create documents, not instantiate them directly. This will make it easy to add new document types in the future without changing the client code.


```python
from abc import ABC, abstractmethod


class Document(ABC):
    @abstractmethod
    def display(self):
        ...


class PDFDocument(Document):
    def display(self):
        print("Отображаю PDF-документ")


class WordDocument(Document):
    def display(self):
        print("Отображаю Word-документ")


class WebDocument(Document):
    def display(self):
        print("Отображаю Web-документ")


class DocumentFactory:
    _registry = {
        "pdf": PDFDocument,
        "word": WordDocument,
        "web": WebDocument,
    }

    @staticmethod
    def create_document(doc_type):
        cls = DocumentFactory._registry.get(doc_type.lower())
        if cls is None:
            raise ValueError(
                f"Неизвестный тип документа: {doc_type!r}. "
                f"Доступно: {', '.join(DocumentFactory._registry)}"
            )
        return cls()

    @staticmethod
    def register(doc_type, cls):
        DocumentFactory._registry[doc_type.lower()] = cls
```

## Задание 1: Счетчик слов

Напишите функцию, которая читает текстовый файл и возвращает словарь с количеством:
- строк
- слов
- символов (без пробелов)

**Файл:** `data/sample_text.txt`

**Требования:**
- Обработайте исключение `FileNotFoundError`
- Используйте `encoding='utf-8'`
- Верните словарь с ключами: `'lines'`, `'words'`, `'chars'`

```python
def count_file_stats(filepath):
    try:
        with open(filepath, 'r', encoding='utf-8') as f:
            content = f.read()
    except FileNotFoundError:
        print(f"Файл не найден: {filepath}")
        return None
    
    return {
        'lines': len(content.splitlines()),
        'words': len(content.split()),
        'chars': len(content.replace(' ', '')),
    }


if __name__ == '__main__':
    result = count_file_stats('data/sample_text.txt')
    print(result)
```

```python
from pathlib import Path


def save_list_to_file(items, filepath):
    path = Path(filepath)
    
    path.parent.mkdir(parents=True, exist_ok=True)
    
    with open(path, 'w', encoding='utf-8') as f:
        for item in items:
            f.write(item + '\n')
```

## Задание 3: Поиск в файле

Напишите функцию, которая ищет в текстовом файле все строки, содержащие заданное слово (регистронезависимо), и возвращает список этих строк.

**Файл:** `data/sample_text.txt`

**Требования:**
- Поиск должен быть регистронезависимым
- Верните список найденных строк
- Обработайте `FileNotFoundError`

```python
def find_lines_with_word(filepath, word):

    try:
        with open(filepath, 'r', encoding='utf-8') as f:
            lines = f.readlines()
    except FileNotFoundError:
        print(f"Файл не найден: {filepath}")
        return None
    
    word_lower = word.lower()
    return [line.rstrip('\n') for line in lines if word_lower in line.lower()]


if __name__ == '__main__':
    filepath = 'data/sample_text.txt'
    
    word = input("Введи слово для поиска: ").strip()
    
    if not word:
        print("Слово не введено")
    else:
        result = find_lines_with_word(filepath, word)
        
        if result is None:
            print("Файл не найден")
        elif result:
            print(f"\nНайдено строк: {len(result)}\n")
            for i, line in enumerate(result, 1):
                print(f"  {i}. {line}")
        else:
            print("Ничего не найдено")
```

## Задание 5: Объединение файлов

Напишите функцию, которая объединяет несколько текстовых файлов в один. Перед содержимым каждого файла добавьте заголовок вида `=== имя_файла.txt ===`.

**Файлы:** `data/file1.txt`, `data/file2.txt`, `data/file3.txt`

**Требования:**
- Пропускайте несуществующие файлы
- Верните количество успешно объединенных файлов
- Добавляйте заголовок перед содержимым каждого файла

```python
# task5.py
from pathlib import Path


def merge_files(filepaths, output_path):
    Path(output_path).parent.mkdir(parents=True, exist_ok=True)
    
    merged_count = 0
    
    with open(output_path, 'w', encoding='utf-8') as out:
        for filepath in filepaths:
            path = Path(filepath)
            
            if not path.exists():
                print(f"⚠️  Пропущен (не найден): {filepath}")
                continue
            
            out.write(f"=== {path.name} ===\n")
            content = path.read_text(encoding='utf-8')
            out.write(content)
            
            if not content.endswith('\n'):
                out.write('\n')
            
            merged_count += 1
    
    return merged_count


if __name__ == '__main__':
    files = [
        'data/file1.txt',
        'data/file2.txt',
        'data/file3.txt',
    ]
    output = 'data/merged.txt'
    
    print(f"Текущая папка: {Path.cwd()}")
    print(f"Объединяю: {files}\n")
    
    count = merge_files(files, output)
    
    print(f"\n Объединено файлов: {count}")
    print(f"Результат сохранён: {output}\n")
    
    print("=" * 60)
    print("Содержимое data/merged.txt:")
    print("=" * 60)
    print(Path(output).read_text(encoding='utf-8'))
```

## Задание 6: Фильтрация CSV

Напишите функцию, которая фильтрует строки CSV файла по заданному условию и сохраняет результат в новый CSV файл.

**Файл:** `data/employees.csv`

**Требования:**
- Используйте `csv.DictReader` и `csv.DictWriter`
- Третий параметр - функция-фильтр (lambda)
- Верните количество строк, прошедших фильтр
- **ОБЯЗАТЕЛЬНО:** Напишите тесты для этой функции


```python
import csv
from pathlib import Path


def filter_csv(input_path, output_path, filter_func):
    try:
        with open(input_path, 'r', encoding='utf-8', newline='') as f_in:
            reader = csv.DictReader(f_in)
            fieldnames = reader.fieldnames
            rows = [row for row in reader if filter_func(row)]
    except FileNotFoundError:
        print(f" Файл не найден: {input_path}")
        return 0
    
    Path(output_path).parent.mkdir(parents=True, exist_ok=True)
    
    with open(output_path, 'w', encoding='utf-8', newline='') as f_out:
        writer = csv.DictWriter(f_out, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(rows)
    
    return len(rows)

if __name__ == '__main__':
    input_file = 'data/employees.csv'
    output_file = 'data/employees_filtered.csv'
    
    if Path(input_file).exists():
        with open(input_file, 'r', encoding='utf-8', newline='') as f:
            reader = csv.DictReader(f)
            print(f" Файл: {input_file}")
            print(f"Заголовки: {reader.fieldnames}")
            rows = list(reader)
            print(f"Всего строк: {len(rows)}\n")
            print("Первые 3 строки:")
            for row in rows[:3]:
                print(f"  {row}")
            print()
    else:
        print(f" Файл не найден: {input_file}")
        exit(1)
    
    print("=" * 60)
    print("ЗАПУСК ТЕСТОВ")
    print("=" * 60)
    
    sample = rows[0]
    numeric_col = None
    for col in ['salary', 'age', 'experience', 'years']:
        if col in sample:
            numeric_col = col
            break
    
    if numeric_col is None:
        print("Не нашли числовой столбец для теста. Столбцы:", list(sample.keys()))
        exit(1)
    
    print(f"Будем фильтровать по столбцу: '{numeric_col}'\n")
    
    count = filter_csv(input_file, 'data/test_all.csv', lambda row: True)
    assert count == len(rows), f"ТЕСТ 1 провален: ожидали {len(rows)}, получили {count}"
    print(f" ТЕСТ 1: фильтр 'все строки' → {count} строк")
    
    count = filter_csv(input_file, 'data/test_none.csv', lambda row: False)
    assert count == 0, f"ТЕСТ 2 провален: ожидали 0, получили {count}"
    print(f" ТЕСТ 2: фильтр 'ничего' → {count} строк")
    
    threshold = 50000
    expected = sum(1 for r in rows if int(r[numeric_col]) > threshold)
    count = filter_csv(
        input_file,
        'data/test_filtered.csv',
        lambda row: int(row[numeric_col]) > threshold
    )
    assert count == expected, f"ТЕСТ 3 провален: ожидали {expected}, получили {count}"
    print(f" ТЕСТ 3: {numeric_col} > {threshold} → {count} строк")
    
    with open('data/test_filtered.csv', 'r', encoding='utf-8', newline='') as f:
        result_rows = list(csv.DictReader(f))
    assert len(result_rows) == expected, f"ТЕСТ 4 провален: в файле {len(result_rows)}, ожидали {expected}"
    assert all(int(r[numeric_col]) > threshold for r in result_rows), "ТЕСТ 4: в файле есть строки, не прошедшие фильтр"
    print(f" ТЕСТ 4: содержимое файла соответствует фильтру ({len(result_rows)} строк)")
    
    with open(input_file, 'r', encoding='utf-8', newline='') as f:
        headers_in = csv.DictReader(f).fieldnames
    with open('data/test_filtered.csv', 'r', encoding='utf-8', newline='') as f:
        headers_out = csv.DictReader(f).fieldnames
    assert headers_in == headers_out, f"ТЕСТ 5: заголовки не совпадают: {headers_in} vs {headers_out}"
    print(f" ТЕСТ 5: заголовки сохранены: {headers_out}")
    
    count = filter_csv('data/nonexistent.csv', 'data/x.csv', lambda row: True)
    assert count == 0, f"ТЕСТ 6 провален: ожидали 0, получили {count}"
    print(f" ТЕСТ 6: несуществующий файл → {count}")
    
    str_col = None
    for col in ['department', 'position', 'name', 'city']:
        if col in sample:
            str_col = col
            break
    
    if str_col:
        target = rows[0][str_col]
        expected = sum(1 for r in rows if r[str_col] == target)
        count = filter_csv(
            input_file,
            'data/test_str.csv',
            lambda row: row[str_col] == target
        )
        assert count == expected, f"ТЕСТ 7 провален: ожидали {expected}, получили {count}"
        print(f" ТЕСТ 7: {str_col} == '{target}' → {count} строк")
    
    print("\n" + "=" * 60)
    print(" ВСЕ ТЕСТЫ ПРОЙДЕНЫ")
    print("=" * 60)
    
    print("\nФинальная фильтрация:")
    final_output = 'data/employees_filtered.csv'
    final_count = filter_csv(
        input_file,
        final_output,
        lambda row: int(row[numeric_col]) > 50000
    )
    print(f"Отфильтровано: {final_count} строк")
    print(f"Результат: {final_output}")
```

## Задание 7: Обновление JSON данных

Напишите функцию, которая читает JSON файл с товарами, применяет скидку к товарам дороже заданной цены, и сохраняет результат в новый JSON файл.

**Файл:** `data/products.json`

**Требования:**
- Функция принимает: путь к файлу, минимальную цену, процент скидки, путь для сохранения
- Примените скидку только к товарам с `price >= min_price`
- Сохраните результат с `indent=2` и `ensure_ascii=False`
- Верните количество товаров, к которым была применена скидка
- Обработайте `FileNotFoundError` и `json.JSONDecodeError`

```python
import json
from pathlib import Path


def apply_discount(input_path, min_price, discount_percent, output_path):
    try:
        with open(input_path, 'r', encoding='utf-8') as f:
            data = json.load(f)
    except FileNotFoundError:
        print(f" Файл не найден: {input_path}")
        return 0
    except json.JSONDecodeError as e:
        print(f" Ошибка в JSON: {e}")
        return 0
    
    if isinstance(data, list):
        products = data
        wrapper = None
    elif isinstance(data, dict):
        wrapper = None
        for key, value in data.items():
            if isinstance(value, list):
                wrapper = key
                products = value
                break
        else:
            print(" Не нашли список товаров в JSON")
            return 0
    else:
        print(" Неожиданный формат JSON")
        return 0
    
    discounted_count = 0
    for product in products:
        price = product.get('price')
        if price is not None and price >= min_price:
            product['price'] = round(price * (1 - discount_percent / 100), 2)
            discounted_count += 1
    
    Path(output_path).parent.mkdir(parents=True, exist_ok=True)
    result = data if wrapper is None else {**data, wrapper: products}
    
    with open(output_path, 'w', encoding='utf-8') as f:
        json.dump(result, f, indent=2, ensure_ascii=False)
    
    return discounted_count

if __name__ == '__main__':
    input_file = 'data/products.json'
    output_file = 'data/products_discounted.json'
    
    if not Path(input_file).exists():
        print(f" Файл не найден: {input_file}")
        print(f"Текущая папка: {Path.cwd()}")
        print(f"data/ содержит: {list(Path('data').iterdir())}")
        exit(1)
    
    with open(input_file, 'r', encoding='utf-8') as f:
        data = json.load(f)
    
    print(f" Файл: {input_file}")
    print(f"Тип верхнего уровня: {type(data).__name__}")
    
    if isinstance(data, list):
        print(f"Товаров: {len(data)}")
        print(f"Первый товар: {data[0]}")
    elif isinstance(data, dict):
        print(f"Ключи: {list(data.keys())}")
        for key, value in data.items():
            if isinstance(value, list):
                print(f"Товаров в '{key}': {len(value)}")
                print(f"Первый товар: {value[0]}")
                break
    
    print()
    print("=" * 60)
    print("ПРИМЕНЯЕМ СКИДКУ")
    print("=" * 60)
    
    min_price = 10000
    discount = 15
    
    print(f"Мин. цена: {min_price}")
    print(f"Скидка: {discount}%\n")
    
    count = apply_discount(input_file, min_price, discount, output_file)
    
    print(f" Скидка применена к {count} товарам")
    print(f"Результат: {output_file}\n")
    
    print("=" * 60)
    print("Первые строки результата:")
    print("=" * 60)
    content = Path(output_file).read_text(encoding='utf-8')
    print('\n'.join(content.splitlines()[:20]))
    if len(content.splitlines()) > 20:
        print(f"... (всего {len(content.splitlines())} строк)")
```

## Задание 8: Копирование файлов

Напишите функцию, которая копирует файлы определенного расширения из одной директории в другую.

**Требования:**
- Используйте `pathlib.Path` и `.glob()`
- Используйте `shutil.copy()`
- Создавайте целевую директорию, если её нет
- Верните список имен скопированных файлов

```python
import shutil
from pathlib import Path


def copy_files_by_extension(source_dir, target_dir, extension):

    source = Path(source_dir)
    target = Path(target_dir)
    
    target.mkdir(parents=True, exist_ok=True)
    
    if not extension.startswith('.'):
        extension = '.' + extension
    pattern = f'*{extension}'
    
    copied = []
    for file_path in source.glob(pattern):
        if file_path.is_file():
            shutil.copy(file_path, target / file_path.name)
            copied.append(file_path.name)
    
    return copied

if __name__ == '__main__':
    source = 'data'
    target = 'data/copied_txt'
    ext = '.txt'
    
    print(f" Источник:  {source}")
    print(f" Назначение: {target}")
    print(f" Расширение: {ext}\n")
    
    print("Содержимое data/:")
    for p in sorted(Path(source).iterdir()):
        marker = "файл" if p.is_file() else "папка"
        print(f"  {marker} {p.name}")
    print()
    
    matching = [p.name for p in Path(source).glob(f'*{ext}') if p.is_file()]
    print(f"Файлы с расширением '{ext}': {matching}\n")
    
    print("=" * 60)
    print("КОПИРУЕМ")
    print("=" * 60)
    
    copied = copy_files_by_extension(source, target, ext)
    
    print(f" Скопировано файлов: {len(copied)}\n")
    for i, name in enumerate(sorted(copied), 1):
        print(f"  {i}. {name}")
    
    print()
    print("=" * 60)
    print(f"Содержимое {target}:")
    print("=" * 60)
    for p in sorted(Path(target).iterdir()):
        size = p.stat().st_size
        print(f"  {p.name}  ({size} байт)")
```

## Задание 9: Анализатор логов

Напишите две функции для анализа файла логов.

**Формат:** `[2025-10-21 10:15:30] INFO: Application started`

**Функция 1:** Подсчитывает количество записей каждого уровня (INFO, WARNING, ERROR)
**Функция 2:** Находит все записи с уровнем ERROR

**Файл:** `data/application.log`

**Требования:**
- Используйте регулярные выражения (модуль `re`)
- **ОБЯЗАТЕЛЬНО:** Напишите тесты для каждой функции

```python
import re
from pathlib import Path


LOG_PATTERN = re.compile(
    r'^\[(\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2})\]\s+(\w+):\s*(.*)$'
)


def count_log_levels(filepath):
    counts = {'INFO': 0, 'WARNING': 0, 'ERROR': 0}
    
    try:
        with open(filepath, 'r', encoding='utf-8') as f:
            for line in f:
                match = LOG_PATTERN.match(line.strip())
                if match:
                    level = match.group(2)
                    if level in counts:
                        counts[level] += 1
    except FileNotFoundError:
        print(f" Файл не найден: {filepath}")
    
    return counts


def find_error_lines(filepath):
    errors = []
    
    try:
        with open(filepath, 'r', encoding='utf-8') as f:
            for line in f:
                match = LOG_PATTERN.match(line.strip())
                if match and match.group(2) == 'ERROR':
                    errors.append(line.strip())
    except FileNotFoundError:
        print(f" Файл не найден: {filepath}")
    
    return errors


if __name__ == '__main__':
    filepath = 'data/application.log'
    
    if not Path(filepath).exists():
        print(f" Файл не найден: {filepath}")
        print(f"Текущая папка: {Path.cwd()}")
        exit(1)
    
    print("=" * 60)
    print(f"Первые 5 строк {filepath}:")
    print("=" * 60)
    with open(filepath, 'r', encoding='utf-8') as f:
        for i, line in enumerate(f):
            if i >= 5:
                break
            print(f"  {line.rstrip()}")
    print()
    
    print("=" * 60)
    print("ПОДСЧЁТ УРОВНЕЙ")
    print("=" * 60)
    counts = count_log_levels(filepath)
    for level, count in counts.items():
        print(f"  {level:8} : {count}")
    total = sum(counts.values())
    print(f"  {'ВСЕГО':8} : {total}")
    print()
    
    print("=" * 60)
    print("ЗАПИСИ С УРОВНЕМ ERROR")
    print("=" * 60)
    errors = find_error_lines(filepath)
    if errors:
        for i, line in enumerate(errors, 1):
            print(f"  {i}. {line}")
    else:
        print("  (нет)")
    print(f"\nВсего ошибок: {len(errors)}")
```

- Сохраните JSON с `indent=2` и `ensure_ascii=False`
- **ОБЯЗАТЕЛЬНО:** Напишите тесты для функций

```python
import csv
import json
from pathlib import Path


def convert_value(value):

    if not isinstance(value, str):
        return value
    
    stripped = value.strip()
    
    if stripped == '':
        return None
    
    low = stripped.lower()
    if low == 'true':
        return True
    if low == 'false':
        return False
    
    try:
        return int(stripped)
    except ValueError:
        pass
    
    try:
        return float(stripped)
    except ValueError:
        pass
    
    return stripped


def csv_to_json(csv_path, json_path):

    try:
        with open(csv_path, 'r', encoding='utf-8', newline='') as f:
            reader = csv.DictReader(f)
            rows = [
                {key: convert_value(val) for key, val in row.items()}
                for row in reader
            ]
    except FileNotFoundError:
        print(f" Файл не найден: {csv_path}")
        return 0
    
    Path(json_path).parent.mkdir(parents=True, exist_ok=True)
    
    with open(json_path, 'w', encoding='utf-8') as f:
        json.dump(rows, f, indent=2, ensure_ascii=False)
    
    return len(rows)

def run_tests():
    print("=" * 60)
    print("ЗАПУСК ТЕСТОВ")
    print("=" * 60)
    
    assert convert_value('') is None
    assert convert_value('   ') is None
    print(" ТЕСТ 1: пустая строка → None")
    
    assert convert_value('true') is True
    assert convert_value('false') is False
    assert convert_value('True') is True
    assert convert_value('FALSE') is False
    assert convert_value('TrUe') is True
    print(" ТЕСТ 2: true/false → bool (регистр не важен)")
    
    assert convert_value('42') == 42
    assert convert_value('-7') == -7
    assert convert_value('0') == 0
    assert isinstance(convert_value('42'), int)
    print(" ТЕСТ 3: целые числа → int")
    
    assert convert_value('3.14') == 3.14
    assert convert_value('-0.5') == -0.5
    assert isinstance(convert_value('3.14'), float)
    print(" ТЕСТ 4: дробные числа → float")
    
    assert convert_value('hello') == 'hello'
    assert convert_value('  hello  ') == 'hello'   # strip
    assert convert_value('12abc') == '12abc'       # не число
    assert convert_value('1,5') == '1,5'           # запятая не парсится
    print(" ТЕСТ 5: не-числа остаются строками")
    
    assert convert_value(None) is None
    assert convert_value(42) == 42
    assert convert_value(True) is True
    print(" ТЕСТ 6: не-строки не трогаем")
    
    test_csv = Path('data/_test_mixed.csv')
    test_csv.write_text(
        "id,name,age,active,score,note\n"
        "1,Alice,30,true,95.5,hello\n"
        "2,Bob,,false,88,world\n"
        "3,,25,TRUE,,\n",
        encoding='utf-8'
    )
    
    test_json = Path('data/_test_mixed.json')
    count = csv_to_json(test_csv, test_json)
    
    assert count == 3, f"ТЕСТ 7: ожидали 3, получили {count}"
    
    with open(test_json, 'r', encoding='utf-8') as f:
        data = json.load(f)
    
    assert data[0] == {'id': 1, 'name': 'Alice', 'age': 30, 'active': True, 'score': 95.5, 'note': 'hello'}
    assert data[1] == {'id': 2, 'name': 'Bob', 'age': None, 'active': False, 'score': 88, 'note': 'world'}
    assert data[2] == {'id': 3, 'name': None, 'age': 25, 'active': True, 'score': None, 'note': None}
    print(" ТЕСТ 7: csv_to_json конвертирует всё правильно")
    
    with open(test_json, 'r', encoding='utf-8') as f:
        data = json.load(f)
    
    assert isinstance(data[0]['id'], int)
    assert isinstance(data[0]['active'], bool)
    assert isinstance(data[0]['score'], float)
    assert data[1]['age'] is None
    assert isinstance(data[1]['score'], int)   # 88 → int, не float
    print(" ТЕСТ 8: типы в JSON правильные (int/bool/float/None)")
    
    test_ru = Path('data/_test_ru.csv')
    test_ru.write_text(
        "name,city\n"
        "Иван,Москва\n"
        "Мария,Санкт-Петербург\n",
        encoding='utf-8'
    )
    test_ru_json = Path('data/_test_ru.json')
    csv_to_json(test_ru, test_ru_json)
    
    content = test_ru_json.read_text(encoding='utf-8')
    assert 'Иван' in content, "ТЕСТ 9: русские буквы должны быть как есть"
    assert 'Москва' in content, "ТЕСТ 9"
    assert '\n  ' in content, "ТЕСТ 9: должен быть отступ в 2 пробела"
    print(" ТЕСТ 9: ensure_ascii=False + indent=2 работают")
    
    count = csv_to_json('data/_nonexistent.csv', 'data/_x.json')
    assert count == 0, f"ТЕСТ 10: ожидали 0, получили {count}"
    print(" ТЕСТ 10: несуществующий файл → 0")
    
    empty_csv = Path('data/_empty.csv')
    empty_csv.write_text("id,name\n", encoding='utf-8')
    count = csv_to_json(empty_csv, 'data/_empty.json')
    assert count == 0
    with open('data/_empty.json', 'r', encoding='utf-8') as f:
        assert json.load(f) == []
    print(" ТЕСТ 11: CSV только с заголовками → пустой список")
    
    for f in [test_csv, test_json, test_ru, test_ru_json, empty_csv,
              Path('data/_empty.json')]:
        if f.exists():
            f.unlink()
    
    print("\n" + "=" * 60)
    print(" ВСЕ ТЕСТЫ ПРОЙДЕНЫ")
    print("=" * 60)


if __name__ == '__main__':
    run_tests()
    print()
    
    csv_path = 'data/mixed_data.csv'
    json_path = 'data/mixed_data.json'
    
    if not Path(csv_path).exists():
        print(f" Файл не найден: {csv_path}")
        print(f"Текущая папка: {Path.cwd()}")
        exit(1)
    
    print("=" * 60)
    print(f"Исходный {csv_path}:")
    print("=" * 60)
    with open(csv_path, 'r', encoding='utf-8', newline='') as f:
        reader = csv.DictReader(f)
        print(f"Заголовки: {reader.fieldnames}")
        for i, row in enumerate(reader):
            if i >= 3:
                break
            print(f"  {row}")
    print()
    
    print("=" * 60)
    print("КОНВЕРТАЦИЯ")
    print("=" * 60)
    count = csv_to_json(csv_path, json_path)
    print(f" Конвертировано записей: {count}")
    print(f"Результат: {json_path}\n")
    
    print("=" * 60)
    print("Первые строки JSON:")
    print("=" * 60)
    content = Path(json_path).read_text(encoding='utf-8')
    print('\n'.join(content.splitlines()[:25]))
    if len(content.splitlines()) > 25:
        print(f"... (всего {len(content.splitlines())} строк)")
```
